# 02 — Interactive Pipeline Testing

**Purpose:** exercise the full `Speech -> STT -> Translation -> TTS` pipeline interactively,
without needing to launch Streamlit. Good for quick manual checks while developing —
try a new audio clip, inspect intermediate outputs, listen to the result, without a full app reload.

# Setup


```python

import sys
sys.path.append("..")

from src.pipeline import run_pipeline
from IPython.display import Audio, display
import time
```

## Run the full pipeline: Nepali → English

Point `AUDIO_PATH` at a real Nepali speech clip.


```python
AUDIO_PATH = "../src/test_ne.mp3"

start = time.time()
result = run_pipeline(AUDIO_PATH, source_lang="ne", target_lang="en")
elapsed = time.time() - start

print(f"Recognized (NE): {result['recognized_text']}")
print(f"Translated (EN): {result['translated_text']}")
print(f"Total time: {elapsed:.2f}s")

display(Audio(result["output_audio_path"]))
```

## Run the full pipeline: English → Nepali

Point `AUDIO_PATH` at a real English speech clip.


```python
AUDIO_PATH = "../src/test_en.mp3"
start = time.time()
result = run_pipeline(AUDIO_PATH, source_lang="en", target_lang="ne")
elapsed = time.time() - start

print(f"Recognized (EN): {result['recognized_text']}")
print(f"Translated (NE): {result['translated_text']}")
print(f"Total time: {elapsed:.2f}s")

display(Audio(result["output_audio_path"]))
```
